# 01 - Data Understanding

Dataset: `diabetes.csv` (as used by the base paper — Kaggle `johndasilva/diabetes`, labeled "Frankfurt Hospital Diabetes Dataset"). Goal: understand structure, quality, and target balance before any modeling.

```
import pandas as pd
from pathlib import Path

DATA_PATH = Path("../data/raw/diabetes.csv")
OUTPUT_PATH = Path("../outputs/reports")
OUTPUT_PATH.mkdir(parents=True, exist_ok=True)

df = pd.read_csv(DATA_PATH)
print("Dataset loaded successfully.")
```

**Missing values (explicit NaNs) and duplicate rows**

Note: this dataset encodes missing clinical values as `0` (e.g. Glucose=0, BMI=0 are physiologically impossible) rather than NaN. We check both.

```
print("Explicit NaNs:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

zero_check_cols = ["Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI"]
print("\nZeros in clinically-impossible columns (likely disguised missing values):")
print((df[zero_check_cols] == 0).sum())
```

**Target distribution**

```
print(df["Outcome"].value_counts())
print()
print(df["Outcome"].value_counts(normalize=True) * 100)
```

**Note on dataset identity**

This file is structurally identical to the classic Pima Indians Diabetes dataset (768 rows, 9 columns, 500/268 class split, same feature ranges). The base paper labels its source (Kaggle `johndasilva/diabetes`) as the "Frankfurt Hospital Diabetes Dataset," but that Kaggle listing is itself a re-upload of the Pima dataset — it is not the same as the actual Frankfurt (Germany) hospital dataset (which has ~2000 rows). This distinction matters for the comparison in the Model Evaluation notebook: we are benchmarking against the paper's *reported numbers*, on the *same data source* they used, not a different or larger dataset.

```
summary = pd.DataFrame({
    "Column": df.columns,
    "Data Type": df.dtypes.values,
    "Zeros (susp. missing)": [(df[c] == 0).sum() if c in zero_check_cols else "-" for c in df.columns],
    "Unique Values": df.nunique().values
})

summary.to_csv(OUTPUT_PATH / "dataset_summary.csv", index=False)
summary
```
